Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

### Ways to get the Viewer to First View the first 3 Months:
* To get the viewer to first view the last 3 months, you could create two seperate graphs, one with the latter 3 months, the latter with the first 3 months and place the first atop of the other.
* Additionally, I could use different colors to seperate out the months, having a more eye-catching color for the last 3 months to draw the user's attention there.
* I could also use a thicker line for the last 3 months to draw attention to them.
* I could also use markers on the last 3 line to make them pop out more to the eye.
* Lastly, I could use a background shading behind the last 3 months to make them pop out more.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt

# Create Dummy Data
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]

received = [160, 185, 240, 145, 180, 155, 130, 202, 160, 139, 149, 177]
processed = [160, 185, 235, 145, 180, 150, 122, 156, 126, 104, 124, 140]
```

```python
plt.figure(figsize=(10,6))

# Plot lines
plt.plot(months, received, color="#777777", marker="o", linewidth=2.5, label="Received")
plt.plot(months, processed, color="#1f4e79", marker="o", linewidth=2.5, label="Processed")

# Titles and labels
plt.title("Ticket volume over time", fontsize=14, weight="bold")
plt.ylabel("Number of tickets")

# Grid and limits
plt.ylim(0, 300)
plt.grid(axis="y", linestyle="--", alpha=0.4)

# Annotate values (Aug–Dec like in the graphic)
labels_received = {7:202, 8:160, 9:139, 10:149, 11:177}
labels_processed = {7:156, 8:126, 9:104, 10:124, 11:140}

for i,v in labels_received.items():
    plt.text(i, v+6, str(v), ha="center", color="#777777")

for i,v in labels_processed.items():
    plt.text(i, v-12, str(v), ha="center", color="#1f4e79")

# Style
plt.legend(loc="upper right")
plt.tight_layout()

# Footnote
plt.figtext(0.01, -0.02, "Data source: XYZ Dashboard, as of 12/31", fontsize=9)

plt.show()
```
